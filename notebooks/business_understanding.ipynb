# Trump Tariffs and International Trade

## 1. Business Understanding

### Project objective
Investigate whether the data support the proposition that Trump's
second-term tariffs reduced US goods imports.

### Investigative questions
1. Was average year-over-year import growth lower during Trump's
   second term than in the preceding period?
2. Were monthly import values lower during the second term after
   accounting for their long-term linear trend?
3. How did the average monthly US goods trade balance differ between
   the second term and the preceding period?

### Scope
Use Census Bureau monthly goods imports, exports, and trade balance
for January 1993–July 2026. Compare January 2025 onward with the
preceding period, which includes Trump's first term.

### Hypotheses
For Questions 1 and 2:
- H₀: The second-term coefficient is greater than or equal to zero.
- H₁: The second-term coefficient is negative.

Question 3 is descriptive: compare average monthly trade balances
and report the difference in dollars.

### Success criteria
Answer each question with a statistic, table, or chart. Report
uncertainty for the regressions and distinguish period comparisons
from causal tariff effects.

### Limitations
The presidential indicator does not isolate tariff effects or provide
a verified no-tariff counterfactual. Only 19 second-term months are
available. Dollar values are not inflation-adjusted, and trade-balance
comparisons may reflect long-term trends and other economic influences.
