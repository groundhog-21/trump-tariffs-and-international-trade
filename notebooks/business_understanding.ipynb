# Trump Tariffs and International Trade

## 1. Business Understanding

### Project objective
Investigate whether Trump’s second-term tariffs reduced US goods imports.

### Research question
Were imports lower than they would have been without those tariffs?

### Scope
Use Census Bureau monthly goods imports, January 1993–July 2026.
January 2025 marks the second term; tariff implementation dates
must be distinguished from presidential tenure.

### Hypotheses
H₀: Tariffs did not reduce imports relative to the no-tariff counterfactual.
H₁: Tariffs reduced imports relative to that counterfactual.

### Success criteria
Estimate the import difference against a defensible counterfactual,
report uncertainty, and account for temporal dependence.

### Limitations
Our presidential dummy alone cannot identify tariff effects.
Slower YoY growth does not necessarily mean lower import levels.
Only 19 second-term months are available; aggregate import values
also reflect prices, exchange rates, demand, and other policies.
